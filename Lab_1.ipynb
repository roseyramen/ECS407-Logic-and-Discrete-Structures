```python
####################### DO NOT CHANGE THESE ##############################
propositions = [(True, True), (True, False), (False, True), (False, False)]

print("Example: AND")
for proposition in propositions:
    A, B = proposition
    statement = A and B
    print("A: {:^10}B: {:^10}=> {:^10}".format(A, B, statement))

print("Example: OR")
for proposition in propositions:
    A, B = proposition
    statement = A or B
    print("A: {:^10}B: {:^10}=> {:^10}".format(A, B, statement))

print("Example: NOT A")
for proposition in [True, False]:
    A = proposition
    statement = not A
    print("A: {:^10} => {:^10}".format(A, statement))

###########################################################################
```
